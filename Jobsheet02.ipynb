C. Menggunakan Fungsi MATH

```python
import numpy as np

# Matriks
A = np.array([[80, 30, 25],
              [50, 20, 10]])

B = np.array([[30, 60, 15],
              [10, 30, 30]])

hasil = A + B
print(hasil)

# Dot product
A = np.array([10, 20, 15])
B = np.array([16, 67, 35])

hasil = np.dot(A, B)
print(hasil)

# Cross product
hasil = np.cross(A, B)
print(hasil)
```

**1. Tentukan hasil perkalian berikut menggunakan kode python**

```python
import numpy as np

A = np.array([[1, 2],
              [3, 4]])

B = np.array([[5, 6],
              [7, 8]])

C = np.array([[1, 0],
              [0, 1]])

hasil1 = np.dot(A, B)
hasil2 = np.dot(B, C)

print("A x B =")
print(hasil1)

print("B x C =")
print(hasil2)
```

**3.	Dua Buah Vektor A dan B merupakan vektor 3D pada koordinat kartesian. Jika Vektor A = 5i + 3j + 7k dan Vektor B = 12i – 3j + k. Maka tentukan panjang masing-masing Vektor, hasil perkalian titik A.B, dan AxB.**

```python
import numpy as np

print("Panjang Vektor")
#Panjang Vektor
A = np.array([5, 3, 7])
B = np.array([12, -3, 1])

panjang_A = np.linalg.norm(A)
panjang_B = np.linalg.norm(B)

print(panjang_A)
print(panjang_B)

print("Dot Product")
#B. Dot Prodyct
dot = np.dot(A, B)
print(dot)

print("Cross Product")
#Cross Product (A x B)
cross = np.cross(A, B)
print(cross)
```
