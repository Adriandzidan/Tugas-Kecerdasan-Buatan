# **LATIHAN**

**A.Central Tendency**

```python
import numpy as np
from scipy import stats

score = [90, 80, 70, 65, 110, 80, 83, 67, 94]

# Mean
mean = np.mean(score)

# Median
median = np.median(score)

# Mode
mode = stats.mode(score, keepdims=True)

print(mean, median, mode)
```

**B.Spread**

```python
import numpy as np

data = [2, 3, 7, 8, 9, 12, 15, 20, 50, 42]

# Range
range_val = np.max(data) - np.min(data)

# Percentile
p90 = np.percentile(data, 90)

# Quartile & IQR
data2 = [3, 4, 4, 5, 5, 6, 6, 7, 8, 8, 9, 10]

Q1 = np.percentile(data2, 25)
Q3 = np.percentile(data2, 75)
IQR = Q3 - Q1

# Variance & Std
data3 = [86, 87, 88, 86, 87, 85, 86]
var = np.var(data3)
std = np.std(data3)

print(range_val, p90, Q1, Q3, IQR, var, std)
```

# **TUGAS PRAKTIKUM**

1.	Pelemparan dadu sebanyak 25 kali. Angka yang keluar datanya adalah :

```python
import numpy as np

data = [2, 3, 7, 8, 9, 12, 15, 20, 42, 50]

# Urutkan Data
print("Urutkan Data")
data_sorted = sorted(data)
print(data_sorted)

# Hitung Q1, Q3, IQR
print("\nHitung Q1, Q3, IQR")
Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)
IQR = Q3 - Q1

print("Q1 =", Q1)
print("Q3 =", Q3)
print("IQR =", IQR)

# Outlier
print("\nOutlier")
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print("Lower Bound =", lower)
print("Upper Bound =", upper)

outlier = [x for x in data if x < lower or x > upper]
print("Data Outlier =", outlier)
```

2.	**Diketahui sebuah deret data 9, 10, 11, 6, 8, 7, 7, 5, 4, 5. Tentukan persentil ke-25 dan persentil ke-75 !**

```python
import numpy as np

# Data nilai
data = [9, 10, 11, 6, 8, 7, 7, 5, 4, 5]

# Mengurutkan data (opsional, tapi bagus untuk melihat posisi data)
data_sorted = sorted(data)
print("Data setelah diurutkan:", data_sorted)

# Menghitung persentil
p25 = np.percentile(data, 25)   # Persentil ke-25 (Q1)
p75 = np.percentile(data, 75)   # Persentil ke-75 (Q3)

# Menampilkan hasil
print("\nHasil Perhitungan:")
print("Persentil ke-25 (P25 / Q1) =", p25)
print("Persentil ke-75 (P75 / Q3) =", p75)
```

3. **Analisis Nilai Ujian**

```python
import numpy as np

print("Program mulai...")

# Data nilai
data = [50, 54, 62, 50, 52, 59, 61, 63, 65, 10, 53, 63, 65, 50, 59, 62, 50, 51, 57, 60, 63, 65, 65, 53, 99]

# a. Urutkan data
data_sorted = sorted(data)
print("\na. Data Terurut:")
print(data_sorted)

# b. Q1, Q3, IQR
Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)
IQR = Q3 - Q1

print("\nb. Quartile & IQR:")
print("Q1 =", Q1)
print("Q3 =", Q3)
print("IQR =", IQR)

# c. Upper & Lower Bound
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print("\nc. Batas Outlier:")
print("Lower Bound =", lower)
print("Upper Bound =", upper)

# d. Outlier
outlier = [x for x in data if x < lower or x > upper]

print("\nd. Data Outlier:")
print(outlier)

print("\nProgram selesai ✅")
```

4. **Standar Deviasi (Data Kelompok)**

```python
import numpy as np

mid = np.array([43, 48, 52.5, 58, 63, 68])
freq = np.array([2, 3, 6, 4, 5, 2])

mean = np.sum(mid * freq) / np.sum(freq)

var = np.sum(freq * (mid - mean)**2) / np.sum(freq)
std = np.sqrt(var)

print(std)
```
